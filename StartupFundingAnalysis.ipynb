```python
# drop remarks column
df.drop(columns=['Remarks'], inplace=True)
```

```python
# set sr no to default index
df.set_index('Sr No', inplace=True)
```

```python
df.columns
```

```python
# changing columns name
df.columns=['date', 'startup', 'vertical', 'subvertical', 'city', 'inverstors', 'inv_types', 'amount']
```

```python
df.columns
```

```python
df.head()
```

```python
df['amount'].isnull().value_counts()
```

```python
df[df['amount'].isnull()]
```

```python
# replacing nan to zero in amount column
df['amount'] = df['amount'].replace(np.nan, '0')
```

```python
df['amount'].isnull().sum()
```

```python
# changing dtype of amount col
df['amount'] = df['amount'].str.replace(',','')
df['amount'] = df['amount'].str.replace('undisclosed','0')
df['amount'] = df['amount'].str.replace('0u0n0k0n0o0w0n0','0')
df['amount'] = df['amount'].str.replace('0U0n0d0i0s0c0l0o0s0e0d0','0')

df = df[df['amount'].str.isdigit()]
df['amount'] = df['amount'].astype(np.float64)
```

```python
df.info()
```

```python
# converting usd to inr(cr)
def convert_to_inr(usd):
    return (usd*95.2) / 10000000
df['amount'] = df['amount'].apply(convert_to_inr)
```

```python
# converting date from str to datetime
pd.to_datetime(df['date'], format='%d/%m/%y', errors='coerce')
```
