```python
import plotly.graph_objects as go
import plotly.express as px
import pandas as pd
import numpy as np
```

```python
import plotly.io as pio
pio.renderers.default = "notebook"
```

```python
iris = px.data.iris()
tips = px.data.tips()
gap = px.data.gapminder()
```

```python
gap.sample()
```

```python
gap_2007 = gap[gap['year'] == 2007]
```

```python
# Scatter plot(using go)
trace1 = go.Scatter(x=gap_2007['lifeExp'], y=gap_2007['gdpPercap'], mode='markers')
data = [trace1]

layout = go.Layout(title='GDP vs Life Exp in 2007', xaxis={'title':'Life Exp'}, yaxis={'title':'GDP'})

fig = go.Figure(data, layout)
fig.show()
```

## Plotly Express

```python
# using express
px.scatter(gap_2007, x='lifeExp', y='gdpPercap', color='continent', size='pop', hover_name='country', size_max=100)
```

```python
# animated graph
px.scatter(gap, x='lifeExp', y='gdpPercap', color='continent', size='pop', hover_name='country', size_max=100,
          animation_frame='year', animation_group='country',
           range_x=(30, 90)
          )
```

```python
# line plot
# india population plot
gap_india = gap[gap['country']=='India']
px.line(gap_india, x='year', y='pop', title='India Population Growth')
```

```python
# plot India, China, US pop
temp_df = gap[gap['country'].isin(['India', 'China', 'United States'])].pivot(index='year', columns='country',values='pop')
px.line(temp_df, x=temp_df.index, y=temp_df.columns, range_x=(1950, 2010), title='China - India - US Population Growth')
```

```python
# gdp
gap['GDP'] = gap['pop']*gap['gdpPercap']
temp_df = gap[gap['country'].isin(['India', 'China', 'United States'])].pivot(index='year', columns='country',values='GDP')
px.line(temp_df, x=temp_df.index, y=temp_df.columns, title='China - India - US GDP')
```

```python
# bar chart
# india pop over years
px.bar(gap_india, x='year', y='pop')
```

```python
# India-China-US group bar chart of pop
temp_df = gap[gap['country'].isin(['India', 'China', 'United States'])].pivot(index='year', columns='country', values='pop')
px.bar(temp_df, x=temp_df.index, y=temp_df.columns,barmode='group', log_y=True, text_auto=True) # 'stack', 'group', 'overlay', 'relative'
```

```python
# cotinent population in 2007
continent_pop = gap[gap['year'] == 2007].groupby('continent')['pop'].sum()
px.bar(continent_pop, x=continent_pop.index, y=continent_pop.values, log_y=True, color=continent_pop.index, title='Population on 2007')
```

```python
# stack bar chart on continent pop, per country contribution in 2007
px.bar(gap_2007, x='continent', y='pop', color='country', log_y=True)
```

```python
# stack bar chart on continent pop, per country contribution
px.bar(gap, x='continent', y='pop', color='country', animation_frame='year',animation_group='country', range_y=(0, 4000000000))
```

```python
# histogram
# hist of life expt of all countries in 2007
px.histogram(gap_2007, x='lifeExp', nbins=15, color='continent', text_auto=True)
```

```python
# pie chart pop of asian contries in 2007
temp_df = gap_2007[gap_2007['continent'] == 'Europe']
px.pie(temp_df, values='pop', names='country', hover_name='country', hover_data='gdpPercap')
```

```python
# per country contribuion to world population in 2007
px.sunburst(gap_2007, path=['continent', 'country'], values='pop', hover_data=['gdpPercap', 'lifeExp'], color='lifeExp')
# hiererchial pie chart or nested pie chart
```

```python
# on tips data
px.sunburst(tips, path=['sex', 'smoker', 'day', 'time'], values='total_bill', color='size', hover_name='tip')
```

```python
# Tree Map
px.treemap(gap_2007, path=[px.Constant('World'),'continent', 'country'], values='pop', hover_data=['gdpPercap', 'lifeExp'], color='lifeExp')
```

```python
# heat map
tips_pivot = tips.pivot_table(index='sex', columns='smoker', values='tip', aggfunc='mean')
px.imshow(tips_pivot)
```

```python
# 3D scatter Plot
px.scatter_3d(gap_2007, x='lifeExp', y='gdpPercap', z='pop', log_z=True, color='continent', hover_name='country')
```

```python
# scatter matrix
px.scatter_matrix(gap, dimensions=['year', 'lifeExp', 'pop'])
```
