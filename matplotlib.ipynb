```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Style

```python
plt.style.available
plt.style.use('dark_background')
```

## 2D Plot

```python
rs_vk_score = pd.read_csv('https://drive.google.com/uc?id=1vI6QgVkD4UZHlTEknKDHJGkdwdbmMx__')
rs_vk_score
```

```python
plt.plot(rs_vk_score['index'], rs_vk_score['V Kohli'], color='red', linestyle='dashed')
# linestyle -> solid, dashed, dotted, dashdot
plt.plot(rs_vk_score['index'], rs_vk_score['RG Sharma'], color='#946B63', linewidth=3) # hexcode

plt.title('VK-RS Total Score in IPL')
plt.xlabel('Season')
plt.ylabel('Run Scored')
```

```python
plt.plot(rs_vk_score['index'], rs_vk_score['V Kohli'], marker='D', label='VK')
plt.plot(rs_vk_score['index'], rs_vk_score['RG Sharma'], marker='>', markersize=10, label='RS')

plt.title('VK-RS Total Score in IPL')
plt.xlabel('Season')
plt.ylabel('Run Scored')

# trim
plt.xlim(2010, 2014)
plt.ylim(300, 700)

plt.legend(loc='best')
# location -> 'best', 'upper right', 'upper left', 'lower left', 'lower right', 'right', 'center left', 'center right', 'lower center', 'upper center', 'center'

plt.grid()
plt.show()
```

## Scatter Plot

```python
batsman = pd.read_csv('https://drive.google.com/uc?id=1kJ1v2GDxa5uADs3uproeMp588RtRggJ_')
top_50_batsman = batsman.sort_values('runs', ascending=False).head(50)
```

```python
top_50_batsman
```

```python
plt.scatter(top_50_batsman['avg'], top_50_batsman['strike_rate'], marker='o', color='red', s=top_50_batsman['runs']/50)
# plt.plot(x, y, 'o') faster but limited
plt.title('Avg - StrikeRate')
plt.xlabel('AVG Runs')
plt.ylabel('Strike Rate')

plt.grid()
plt.show()
```

## Bar Plot
mostly numerical vs categorical

```python
children = [10, 30, 22, 19, 5]
colors = ['red', 'blue', 'green', 'orange', 'black']
```

```python
plt.bar(colors, children, color='orange', width=0.2) # barh for horizontal
```

```python
batsman_season_record = pd.read_csv('https://drive.google.com/uc?id=10p6RZu-q5PGXLwsXAfemJbjAIN_A0BGq')
batsman_season_record
```

```python
# Multiple bar plots
plt.bar(np.arange(5) - 0.2, batsman_season_record['2015'], width=0.2, label='2015')
plt.bar(np.arange(5), batsman_season_record['2016'], width=0.2, label='2016')
plt.bar(np.arange(5) + 0.2, batsman_season_record['2017'], width=0.2, label='2017')

plt.xticks(np.arange(5), batsman_season_record['batsman'], rotation=45)
plt.legend()
plt.show()
```

```python
# stack bar chart
plt.bar(batsman_season_record['batsman'], batsman_season_record['2015'], label='2015')
plt.bar(batsman_season_record['batsman'], batsman_season_record['2016'], bottom=batsman_season_record['2015'], label='2016')
plt.bar(batsman_season_record['batsman'], batsman_season_record['2017'], bottom=(batsman_season_record['2015'] + batsman_season_record['2016']), label='2017')

plt.legend()
plt.show()
```

## Pie-Chart

```python
data = [10, 25, 33, 99]
marks = ['ben', 'eng', 'phy', 'math']
plt.pie(data, labels=marks, autopct='%0.2f%%', colors=['blue', 'cyan', 'yellow', 'gray'], explode=[0.1, 0.5, 0, 0.2], shadow=True)

plt.show()
```
